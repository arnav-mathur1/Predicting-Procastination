```python
import pandas as pd
data = pd.read_csv("p2c-attempts-v1.csv")
data2 = data.replace('A01', 1).replace('A02', 2).replace('A03', 3)
print(data2.dtypes)
```

```python
#data3 = data2.sort_values("student")
data3 = data2.sort_values(['student', 'assignment', "attempt"])
data3
```

```python
data3.min()
```

```python
data3.max()
```

```python
# does procrastinate - 1, does not procrastinate - 0
data3["procrastination"] = data3["submission time"] < 0
data4 = data3.replace(True, 1).replace(False, 0)
print(data4.dtypes)
```

```python
data4 = data4.rename(columns={'submission time': 'submission_time'})
data4
```

```python
# iterating the columns
for col in data4.columns:
    print(col)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# Create the features
features = ["assignment", "attempt", "grade", "submission_time"]

# Create the target variable
target = "procrastination"

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(data4[features], data4[target], test_size=0.20)

# Create the logistic regression model
model = LogisticRegression()

# Train the model
model.fit(X_train, y_train)

# Evaluate the model
score = model.score(X_test, y_test)
print(score)
```
