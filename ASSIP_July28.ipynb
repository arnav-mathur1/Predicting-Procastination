```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data for MSE and R-squared values for each model
models = ['Random Forest', 'Support Vector Regression', 'Neural Network']
mse_values = [0.01287053269412072, 0.03972172221648033, 0.01627221463718088]
r_squared_values = [0.8733387337420709, 0.6090912665811126, 0.8398621595742006]

# Create a figure and axis object
fig, ax = plt.subplots()

# Plot the MSE values
ax.bar(np.arange(len(models)), mse_values, width=0.4, align='center', label='MSE', color='blue')

# Plot the R-squared values on the same graph
ax.bar(np.arange(len(models)) + 0.4, r_squared_values, width=0.4, align='center', label='R-squared', color='orange')

# Set x-axis labels and title
ax.set_xticks(np.arange(len(models)) + 0.2)
ax.set_xticklabels(models)
ax.set_xlabel('Model')
ax.set_ylabel('Value')
ax.set_title('MSE and R-squared values for each model')

# Display the values on the bars
for i in range(len(models)):
    ax.text(i, mse_values[i], f'{mse_values[i]:.4f}', ha='center', va='bottom', color='black')
    ax.text(i + 0.4, r_squared_values[i], f'{r_squared_values[i]:.4f}', ha='center', va='bottom', color='black')

# Add a legend
ax.legend()

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data for MSE and R-squared values for each model
models = ['Random Forest', 'Support Vector Regression', 'Neural Network']
mse_values = [0.01287053269412072, 0.03972172221648033, 0.01627221463718088]
r_squared_values = [0.8733387337420709, 0.6090912665811126, 0.8398621595742006]

# Create a figure and axis object with a larger figsize
fig, ax1 = plt.subplots(figsize=(7, 8))

# Plot the MSE values on the first y-axis
ax1.bar(np.arange(len(models)), mse_values, width=0.4, align='center', label='MSE', color='dodgerblue')
ax1.set_xlabel('Model')
ax1.set_ylabel('MSE', color='dodgerblue')

# Plot the R-squared values on the second y-axis
ax2 = ax1.twinx()
ax2.bar(np.arange(len(models)) + 0.4, r_squared_values, width=0.4, align='center', label='R-squared', color='darkorange')
ax2.set_ylabel('R-squared', color='darkorange')

# Set x-axis labels and title
ax1.set_xticks(np.arange(len(models)) + 0.2)
ax1.set_xticklabels(models)
ax1.set_title('MSE and R-squared values for each model')

# Display the values on the bars
for i in range(len(models)):
    ax1.text(i, mse_values[i], f'{mse_values[i]:.4f}', ha='center', va='bottom', color='black')
    ax2.text(i + 0.4, r_squared_values[i], f'{r_squared_values[i]:.4f}', ha='center', va='bottom', color='black')

# Manually create the legend
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left', bbox_to_anchor=(0.7, 1.0))

# Set y-axis ticks and labels for the second y-axis
ax2.set_yticks(np.linspace(ax2.get_yticks()[0], ax2.get_yticks()[-1], len(ax2.get_yticks())))
ax2.set_yticklabels([f'{r:.4f}' for r in ax2.get_yticks()])

# Show the plot
plt.tight_layout()
plt.show()

fig.savefig('fig.png', dpi=300, bbox_inches='tight')
```
